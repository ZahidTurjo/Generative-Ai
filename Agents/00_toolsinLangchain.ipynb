#Custom Tools

```python
from langchain_core.tools import tool
```

```python
@tool
def addition(a:int,b:int)->int:
  """Add two numbers"""
  return a+b
```

```python
result=addition.invoke({
    "a":2,
    "b":3
})
print(result)
```

```python
print(addition.name)
print(addition.description)
print(addition.args)
```

```python
print(addition.args_schema.model_json_schema())
```

#Custom Tools Using StructuredTool

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class AddtionInput(BaseModel):
  a: int= Field(required=True,description="First Number")
  b: int =Field(required=True,description="Second Number")
```

```python
add=AddtionInput(a=2,b=3)
print(add)
```

```python
def Addition(a:int,b:int)->int:
  return a+b
```

```python
addition_tool=StructuredTool.from_function(
    func=Addition,
    name="ADDITIONSS",
    description="Add two Numbers",
    args_schema=AddtionInput

)
```

```python
result=addition_tool.invoke(
    {
        "a":3,
        "b":7
    }
)
print(result)
```

#Custom Tool using BaseMOdel

```python
from langchain_core.tools import BaseTool
from typing import Type
```

```python
class AdditionInput(BaseModel):
  a : int =Field(required=True,description="First Number")
  b : int =Field(required=True,description="Second Number")
```

```python
class AdditionTool(BaseModel):
  name: str="Addition"
  description: str="Add two numbers"

  args_schema : Type[BaseModel]=AdditionInput

  def invoke(self,a: int,b: int)->int:
    return a*b
```

```python
addition_instance = AdditionTool()
result = addition_instance.invoke(a=3, b=4)
print(result)
```

```python
print(addition_instance.name)
print(addition_instance.description)
print(addition_instance.args_schema)
```

ToolKit

```python
from langchain_core.tools import tool

@tool
def add(a: int,b :int)->int:
  """Add two Numbers"""
  return a+b

@tool
def multiply(a:int,b:int)->int:
  """Multiply Two Numbers"""
  return a*b
```

```python
class AgentToolkit:

  def get_tools(self):
    return[add,multiply]
```

```python
toolkit=AgentToolkit()
```

```python
tools=toolkit.get_tools()
tools
```

```python
for tool in tools:
  print(tool.name,"===",tool.description)
  print(tool.invoke({
      "a":3,"b":4
  }))
print()
print()
```
