# StrOutputParser

```python
from google.colab import userdata
GROQ_API_KEY=userdata.get('GROQ_API_KEY')
```

```python
from langchain_groq import ChatGroq

llm=ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=GROQ_API_KEY
)
```

```python
from langchain_core.prompts import PromptTemplate
template1=PromptTemplate(
    template="give me summery about {topic}"
)
prompt1=template1.invoke({"topic":"Black hole"})
prompt1
```

```python
result1=llm.invoke(prompt1)
result1.content
```

```python
print(result1.content)
```

```python
template2=PromptTemplate(
    template="Give me 3 key point from the summary={summary}"
)
prompt2=template2.invoke({'summary':result1.content})
prompt2
```

```python
result2=llm.invoke(prompt2)
print(result2.content)
```

## Using StrOutputParsr

```python
from langchain_core.output_parsers import StrOutputParser

parser=StrOutputParser()

chain=template1 | llm | parser | template2 | llm | parser

result_chain=chain.invoke({"topic":"Machine Learning"})
print(result_chain)
```

#JsonOutputparser

```python
from langchain_core.output_parsers import JsonOutputParser
parser=JsonOutputParser()
parser.get_format_instructions()
```

```python
template3=PromptTemplate(
    template="Give me 3 key points about {topic}\n {format_instructions}",
    partial_variables={"format_instructions":parser.get_format_instructions()}
)
prompt3=template3.invoke({"topic":"ML"})
prompt3
```

```python
res=llm.invoke(prompt3)
print(res.content)
```

```python
parser.parse(res.content)
```

```python
chain= template3 | llm
res=chain.invoke({"topic":"ML"})
res
```

```python
print(res.content)
```

```python
chain= template3 | llm | parser
res=chain.invoke({"topic":"ML"})
res
```

#pydanticOutputParser

```python
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import Field,BaseModel

class KeyPoint(BaseModel):
  point_1: str=Field(description="Give 1 line point")
  point_2: str=Field(description="Give 2 line point")
  point_3: str=Field(description="Give breif summary")
parser_py=PydanticOutputParser(pydantic_object=KeyPoint)
```

```python
template_py=PromptTemplate(
    template="Give me 3 key points about{topic}\n {format_ins} ",
    partial_variables={"format_ins":parser_py.get_format_instructions()}
)
from rich import print
print(template_py)
```

```python
chain_py= template_py | llm | parser_py

res_py=chain_py.invoke({"topic":"Machine Learning"})
res_py
```

```python
print(res_py)
```
