```python
from google.colab import userdata

GROQ_API_KEY=userdata.get('GROQ_API_KEY')
```

```python
history=[
    SystemMessage(content="You are a helpful ai Assistant")
]
```

```python
from langchain_core.messages import HumanMessage,SystemMessage,AIMessage
```

```python
from langchain_groq import ChatGroq

llm=ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=GROQ_API_KEY
)

while True:
  user_query=input("You: ")
  if user_query=="exict":
    break

  history.append(HumanMessage(content=user_query))
  result=llm.invoke(history)
  print(result.content)
  history.append(AIMessage(content=result.content))
```

```python
from rich import print
print(history)
```

## ChatPrompt Template

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
chattemplate=ChatPromptTemplate(
    [
        ('system','you are {domain} expert'),
        ('human','explain about {topic} in 1 line')
    ]
)
```

```python
print(chattemplate)
```

```python
prompt=chattemplate.invoke({
    "domain":"ML",
    "topic":"NLP"
})
print(prompt)
```

```python
response=llm.invoke(prompt)
print(response.content)
```

```python
from langchain_core.prompts import MessagesPlaceholder
```

```python
chattemplate=ChatPromptTemplate(
    [
    ('system','u are a helpful customer support agent'),
    MessagesPlaceholder(variable_name="chat_his"),
    ('human','{user_query}')

    ]
)
```

```python
prompt=chattemplate.invoke({
    "chat_his":chathistory,
    "user_query":"Were is my refund"
})
prompt
```

```python
result=llm.invoke(prompt)
print(result.content)
```
